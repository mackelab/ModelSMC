```python
import os

import svgutils.transform as sg
import yaml
from svgutils.compose import SVG
```

```python
# read panel sizes from yaml file
with open("panel_sizes_cm.yaml", "r") as file:
    panel_sizes = yaml.safe_load(file)
```

```python
# Set positions
x_pos_cm_panel_landscape = 7
y_pos_cm_panel_landscape = 0.0


x_pos_cm_panel_smc_loop = 0.0
y_pos_cm_panel_smc_loop = 2.6

x_pos_cm_panel_problem = 0.1
y_pos_cm_panel_problem = 0.1

scale_panel_landscape = 1
scale_panel_smc_loop = 1
scale_panel_problem = 1


def cm_top_pix(x_cm):
    return x_cm / 2.54 * 96  # 96 dpi


total_width_cm = panel_sizes["total"]["width_cm"]
total_height_cm = panel_sizes["total"]["height_cm"]

# set the number of pixels per cm used for this figure
fig = sg.SVGFigure(width=f"{total_width_cm}cm", height=f"{total_height_cm}cm")

plot_landscape = sg.fromfile(os.path.join("../panels/", "panel_b.svg"))
plot_landscape = plot_landscape.getroot()

plot_smc_loop = sg.fromfile(os.path.join("../panels/", "panel_cd.svg"))
plot_smc_loop = plot_smc_loop.getroot()

plot_problem = sg.fromfile(os.path.join("../panels/", "panel_a.svg"))
plot_problem = plot_problem.getroot()

txt_a = sg.TextElement(5, 15, "a", size=15, weight="bold", font="Arial")
txt_b = sg.TextElement(280, 15, "b", size=15, weight="bold", font="Arial")
txt_c = sg.TextElement(5, 110, "c", size=15, weight="bold", font="Arial")
txt_d = sg.TextElement(130, 110, "d", size=15, weight="bold", font="Arial")


plot_smc_loop.moveto(
    cm_top_pix(x_pos_cm_panel_smc_loop + 0.25),
    cm_top_pix(y_pos_cm_panel_smc_loop),
    scale_panel_smc_loop,
)


plot_problem.moveto(
    cm_top_pix(x_pos_cm_panel_problem),
    cm_top_pix(y_pos_cm_panel_problem),
    scale_panel_problem,
)

plot_landscape.moveto(
    cm_top_pix(x_pos_cm_panel_landscape),
    cm_top_pix(y_pos_cm_panel_landscape),
    scale_panel_landscape,
)

fig.append([plot_landscape, plot_smc_loop, plot_problem])
fig.append([txt_a, txt_b, txt_c, txt_d])

fig.set_size((f"{total_width_cm}cm", f"{total_height_cm}cm"))
os.makedirs("../fig", exist_ok=True)
fig.save("../fig/fig_1.svg")
svg = SVG("../fig/fig_1.svg")
svg
```
