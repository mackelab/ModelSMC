```python
import os

import matplotlib as mpl
import matplotlib.patches as patches
import matplotlib.pyplot as plt
import yaml

step_font_dict = {"color": "black", "fontfamily": "Arial", "fontsize": 10}
```

```python
with open("panel_sizes_cm.yaml", "r") as file:
    panel_sizes = yaml.safe_load(file)

fig_width_cm = panel_sizes["panel_c"]["width_cm"]
fig_height_cm = panel_sizes["panel_c"]["height_cm"]

scale_panel = 1.25
fig_width_inch = fig_width_cm / 2.54 * scale_panel
fig_height_inch = fig_height_cm / 2.54 * scale_panel
```

```python
with mpl.rc_context(fname="../../.matplotlibrc"):
    fig, ax = plt.subplots(
        figsize=(fig_width_inch, fig_height_inch), layout="constrained"
    )

    ####################################################################################
    # Plot the Default SMC Routine
    ####################################################################################
    # General
    ax.axis("off")
    ax.set_xlim(0, 5.2)
    ax.set_ylim(-2.7, 2.7)

    text = ["resample", "propagate", "weight"]

    dy = 1.2
    x0 = 1.25
    y0 = 0.8

    # Set the title
    ax.text(
        x0, 2.2, "SMC", fontdict=step_font_dict, ha="center", va="center", zorder=11
    )

    # Format of the surounding box
    box_width = 2.2
    box_height = 0.7
    box_lw = 0.5
    num_duplicates = 3
    dx_duplicates = 0.05
    dy_duplicates = 0.05

    # Format of the arrows
    arrow_width = 0.05
    arrow_head_width = 0.15
    arrow_head_length = 0.1
    arrow_lw = 1.0

    for i, t in enumerate(text):
        x_i = x0
        y_i = y0 - i * dy
        ax.text(
            x_i, y_i, t, fontdict=step_font_dict, ha="center", va="center", zorder=11
        )

        rect = patches.Rectangle(
            (x_i - box_width / 2, y_i - box_height / 2),
            box_width,
            box_height,
            fill=True,
            edgecolor="black",
            facecolor="white",
            linewidth=box_lw,
            zorder=10,
        )
        ax.add_patch(rect)

        for i in range(num_duplicates):
            rect_dup = patches.Rectangle(
                (
                    x_i - box_width / 2 + (i + 1) * dx_duplicates,
                    y_i - box_height / 2 + (i + 1) * dy_duplicates,
                ),
                box_width,
                box_height,
                fill=True,
                edgecolor="black",
                facecolor="white",
                linewidth=box_lw,
                zorder=9 - i,
            )
            ax.add_patch(rect_dup)

        # Plot the arrows connecting the boxes
        for i in range(len(text) - 1):
            x_start = x0
            y_start = y0 - i * dy - box_height / 2 + 0.08
            x_end = x0
            y_end = y0 - (i + 1) * dy + box_height / 2 - 0.11

            ax.annotate(
                "",
                xytext=(x_start, y_start),
                xy=(x_end, y_end),
                arrowprops=dict(
                    arrowstyle="->",
                    color="k",
                    linestyle="-",
                    linewidth=arrow_lw,
                    alpha=1.0,
                    zorder=13,
                    mutation_scale=10,
                    capstyle="butt",
                ),
                zorder=11,
            )

        # Plot the iteration arrows
        x_list = [
            x0,
            x0,
            x0 - 1.2,
            x0 - 1.2,
            x0,
            x0,
        ]
        y_list = [
            y0 - 2 * dy - box_height / 2 - 0.017,
            y0 - 3.0 * dy + box_height / 2,
            y0 - 3.0 * dy + box_height / 2,
            y0 + 1.2 * dy - box_height / 2,
            y0 + 1.2 * dy - box_height / 2,
            y0 + 0.8 * dy - box_height / 2,
        ]

        plt.plot(
            x_list,
            y_list,
            marker="",
            linestyle="-",
            color="black",
            linewidth=arrow_lw,
            alpha=1.0,
            zorder=11,
        )

        ax.annotate(
            "",
            xytext=(x0, y0 + 1.2 * dy - box_height / 2 - 0.2),
            xy=(x0, y0 + box_height / 2 - 0.11),
            arrowprops=dict(
                arrowstyle="->",
                color="k",
                linestyle="-",
                linewidth=arrow_lw,
                alpha=1.0,
                zorder=13,
                mutation_scale=10,
                capstyle="butt",
            ),
            zorder=11,
        )

    ####################################################################################
    # Plot ours
    ####################################################################################

    mpl.rcParams["mathtext.fontset"] = "cm"

    x0_b = x0 + 2.65  # 2.64

    # box_width = 1.

    ax.text(
        x0_b,
        2.2,
        "ModelSMC",
        fontdict=step_font_dict,
        ha="center",
        va="center",
        zorder=11,
    )

    text = [
        "resample",
        r"$m_k\sim$ LLM($m_{k-1},c_{k-1})$",
        r"evaluate $p(x_o\mid m_k)$",
    ]

    for i, t in enumerate(text):
        x_i = x0_b
        y_i = y0 - i * dy
        ax.text(
            x_i, y_i, t, fontdict=step_font_dict, ha="center", va="center", zorder=11
        )

        rect = patches.Rectangle(
            (x_i - box_width / 2, y_i - box_height / 2),
            box_width,
            box_height,
            fill=True,
            edgecolor="black",
            facecolor="white",
            linewidth=box_lw,
            zorder=10,
        )
        ax.add_patch(rect)

        for i in range(num_duplicates):
            rect_dup = patches.Rectangle(
                (
                    x_i - box_width / 2 + (i + 1) * dx_duplicates,
                    y_i - box_height / 2 + (i + 1) * dy_duplicates,
                ),
                box_width,
                box_height,
                fill=True,
                edgecolor="black",
                facecolor="white",
                linewidth=box_lw,
                zorder=9 - i,
            )
            ax.add_patch(rect_dup)

        # Plot the arrows connecting the boxes
        for i in range(len(text) - 1):
            x_start = x0_b
            y_start = y0 - i * dy - box_height / 2 + 0.08
            x_end = x0_b
            y_end = y0 - (i + 1) * dy + box_height / 2 - 0.11

            ax.annotate(
                "",
                xytext=(x_start, y_start),
                xy=(x_end, y_end),
                arrowprops=dict(
                    arrowstyle="->",
                    color="k",
                    linestyle="-",
                    linewidth=arrow_lw,
                    alpha=1.0,
                    zorder=13,
                    mutation_scale=10,
                    capstyle="butt",
                ),
                zorder=11,
            )

        # Plot the iteration arrows
        x_list = [
            x0_b,
            x0_b,
            x0_b - 1.2,
            x0_b - 1.2,
            x0_b,
            x0_b,
        ]
        y_list = [
            y0 - 2 * dy - box_height / 2 - 0.017,
            y0 - 3.0 * dy + box_height / 2,
            y0 - 3.0 * dy + box_height / 2,
            y0 + 1.2 * dy - box_height / 2,
            y0 + 1.2 * dy - box_height / 2,
            y0 + 0.8 * dy - box_height / 2,
        ]

        plt.plot(
            x_list,
            y_list,
            marker="",
            linestyle="-",
            color="black",
            linewidth=arrow_lw,
            alpha=1.0,
            zorder=11,
        )

        ax.annotate(
            "",
            xytext=(x0_b, y0 + 1.2 * dy - box_height / 2 - 0.2),
            xy=(x0_b, y0 + box_height / 2 - 0.11),
            arrowprops=dict(
                arrowstyle="->",
                color="k",
                linestyle="-",
                linewidth=arrow_lw,
                alpha=1.0,
                zorder=13,
                mutation_scale=10,
                capstyle="butt",
            ),
            zorder=11,
        )


save_folder = "../panels/"

if not os.path.exists(save_folder):
    os.makedirs(save_folder)
plt.savefig(
    os.path.join(save_folder, "panel_cd.svg"), format="svg", transparent=True, dpi=300
)
plt.close(fig)
```
