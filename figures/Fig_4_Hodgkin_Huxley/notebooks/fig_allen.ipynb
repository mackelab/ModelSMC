```python
import os

import svgutils.transform as sg
import yaml
from IPython.display import SVG

current_dir = os.getcwd()
dir_figures = os.path.join(current_dir, "../", "fig")
dir_panels = os.path.join(current_dir, "../", "panels")
print(dir_figures)
```

```python
with open("panel_sizes_cm.yaml", "r") as file:
    panel_sizes = yaml.safe_load(file)

total_width_cm = panel_sizes["total"]["width_cm"]
total_height_cm = panel_sizes["total"]["height_cm"]
fig = sg.SVGFigure(width=f"{total_width_cm}cm", height=f"{total_height_cm}cm")

plot1 = sg.fromfile(os.path.join(dir_panels, "convergence_allen.svg"))
plot1 = plot1.getroot()

plot2 = sg.fromfile(os.path.join(dir_panels, "model_parts_allen.svg"))
plot2 = plot2.getroot()

plot3 = sg.fromfile(os.path.join(dir_panels, "real_observations_allen.svg"))
plot3 = plot3.getroot()

plot4 = sg.fromfile(os.path.join(dir_panels, "discovered_predictives_allen.svg"))
plot4 = plot4.getroot()

plot5 = sg.fromfile(os.path.join(dir_panels, "baseline_predictives_allen.svg"))
plot5 = plot5.getroot()

txt2 = sg.TextElement(5, 15, "a", size=15, weight="bold", font="Arial")
txt3 = sg.TextElement(340, 15, "c", size=15, weight="bold", font="Arial")
txt1 = sg.TextElement(5, 115, "b", size=15, weight="bold", font="Arial")
plot1.moveto(10, 5, scale=1.0)
plot2.moveto(7, 120, scale=1.0)
plot3.moveto(340, 0, scale=1.0)
plot4.moveto(340, 60, scale=1.0)
plot5.moveto(340, 120, scale=1.0)


fig.append([plot1, plot2, plot3, plot4, plot5])
fig.append([txt1, txt2, txt3])

fig.save(os.path.join(dir_figures, "fig_allen.svg"))

svg = SVG(os.path.join(dir_figures, "fig_allen.svg"))
svg
```
