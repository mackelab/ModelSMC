```python
import os

import svgutils.transform as sg
import yaml
from IPython.display import SVG

current_dir = os.getcwd()
dir_figures = os.path.join("../", "panels")
dir_result = os.path.join("../", "fig")

os.makedirs(dir_result, exist_ok=True)
```

```python
with open("panel_sizes_cm.yaml", "r") as file:
    panel_sizes = yaml.safe_load(file)

total_width_cm = panel_sizes["total"]["width_cm"]
total_height_cm = panel_sizes["total"]["height_cm"]
fig = sg.SVGFigure(width=f"{total_width_cm}cm", height=f"{total_height_cm}cm")

plot1 = sg.fromfile(os.path.join(dir_figures, "resampling_weights_gt_likelihood.svg"))
plot1 = plot1.getroot()

plot2 = sg.fromfile(
    os.path.join(
        dir_figures,
        "gmm_correct_idx_single_run_35e85ba1-0b04-43e8-b95d-cd57e275ef26.svg",
    )
)
plot2 = plot2.getroot()

plot3 = sg.fromfile(
    os.path.join(dir_figures, "ess_35e85ba1-0b04-43e8-b95d-cd57e275ef26.svg")
)
plot3 = plot3.getroot()

scale = 1.4
plot1.moveto(-5 * scale, 6, scale=scale)
plot2.moveto(183 * scale, 0, scale=scale)
plot3.moveto(183 * scale, 85 * scale, scale=scale)

txt1 = sg.TextElement(0, 12, "a", size=15, weight="bold", font="Arial")
txt2 = sg.TextElement(180 * scale, 12, "b", size=15, weight="bold", font="Arial")
txt3 = sg.TextElement(
    180 * scale, 80 * scale, "c", size=15, weight="bold", font="Arial"
)

fig.append([plot1, plot2, plot3])
fig.append([txt1, txt2, txt3])

fig.save(os.path.join(dir_result, "panel_appendix_LLM-free_experiment.svg"))

svg = SVG(os.path.join(dir_result, "panel_appendix_LLM-free_experiment.svg"))
svg
```
